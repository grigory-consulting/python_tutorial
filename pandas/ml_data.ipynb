```python
import kagglehub
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# 1. Download von Kaggle
path = kagglehub.dataset_download("pavansubhasht/ibm-hr-analytics-attrition-dataset")
print("Path to dataset files:", path)

# 2. Laden in ein DataFrame
filepath = os.path.join(path, "WA_Fn-UseC_-HR-Employee-Attrition.csv")
df = pd.read_csv(filepath)
print(df.head())

# 3. Zielvariable numerisch kodieren ('Attrition' -> 1/0)
df['Attrition_Flag'] = (df['Attrition'] == 'Yes').astype(int)

# 4. Features auswählen
features = ['Age', 'MonthlyIncome', 'JobRole', 'DistanceFromHome', 'Gender', 'Department']
X = df[features].copy()
y = df['Attrition_Flag']

# Kategorische Variablen kodieren (LabelEncoder für Demo)
for col in ['JobRole', 'Gender', 'Department']:
    X[col] = LabelEncoder().fit_transform(X[col])

# Optionale Feature-Skalierung (empfohlen für viele ML-Modelle)
num_cols = ['Age', 'MonthlyIncome', 'DistanceFromHome']
scaler = StandardScaler()
X[num_cols] = scaler.fit_transform(X[num_cols])

# 5. Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 6. ML-Modell: Logistische Regression
clf = LogisticRegression(max_iter=200)
clf.fit(X_train, y_train)

# 7. Auswertung
y_pred = clf.predict(X_test)
print("\nML-Ergebnis (Kündigungsprognose Attrition):")
print(classification_report(y_test, y_pred, target_names=['bleibt','kündigt']))
print("Accuracy:", accuracy_score(y_test, y_pred))

# 8. Einzelvorhersage (z.B. 1 Testperson)
print("\nBeispiel-Vorhersage für ersten Test-Datensatz:")
print(X_test.iloc[[0]])
print("Prognose:", clf.predict(X_test.iloc[[0]])[0])
```
