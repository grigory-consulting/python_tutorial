```python
import pandas as pd
from faker import Faker
import random
```

```python
fake = Faker("de_DE")

def random_gender():
    return random.choice(["männlich", "weiblich", 'divers'])

def random_abteilung():
    return random.choice(["IT", "HR", "Vertrieb", "Marketing", "Einkauf"])

def random_contract():
    return random.choice(["befristet", "unbefristet", "azubi"])

def random_tax_class():
    return random.choice(["I", "II", "III", "IV", "V", "VI"])

def random_family_status():
    return random.choice(["ledig", "verheiratet", "geschieden", "verwitwet"])
```

```python
rows = []

for _ in range(30000):
    mitarbeiter = {
        "MitarbeiterID": fake.unique.uuid4(),
        "Dienstnummer": fake.bothify("??#######").upper(), # zwei Mal Buchstabe, 6 mal Zahl
        "Vorname": fake.first_name(),
        "Nachname": fake.last_name(),
        "Geburtsdatum": fake.date_of_birth(minimum_age=18, maximum_age=75),
        "Geschlecht": random_gender(),
        "Adresse": fake.address().replace('\n', ', '), # Bei Adresse die Neuzeilen durch Komma+Leerzeichen ersetzen
        "Telefon": fake.phone_number(),
        "Email": fake.unique.email(),
        "Familienstand": random_family_status(),
        "Beruf": fake.job(),
        "Abteilung":  random_abteilung(),
        "Gehalt_m":random.randint(27000,120000),
        "Vertrag": random_contract(),
        "Letzte Gehaltsverhandlung": fake.date_between(start_date="-5y", end_date="today"), 
        "Steuerklasse": random_tax_class(),
    }
    rows.append(mitarbeiter)

df = pd.DataFrame(rows)
```

# Zweites DataFrame: Fortbildungmaßnahmen

```python
fortbildungs_ids = df["MitarbeiterID"].sample(frac=0.2, random_state = 1)

def random_possibilities():
    return random.choice(["SCRUM", "ISO9001", "Datenschutz", "Leadership", "Python Fortgeschritten", "Java"])

fortb_list = []

for id in fortbildungs_ids:
    eintrag = {
        "MitarbeiterID": id,
        "Fortbildung": random_possibilities(),
        "Dauer_d": random.randint(1,7),
        "Abschluss": random.choice([True, False]),
        "Durchgeführt_am": fake.date_this_year(),
    }
    fortb_list.append(eintrag)

df2 = pd.DataFrame(fortb_list)
df2
```

```python
# 

gemeinsam_df = pd.concat([df,df2], ignore_index=True)

gemeinsam_df
```

```python
# Zusammengeführte Tabelle: Mitarbeiter + Fortbildung

_df = df.set_index("MitarbeiterID")
_df2 = df2.set_index("MitarbeiterID")
gemeinsam_df = pd.concat([_df,_df2], axis=1) 

gemeinsam_df
```

```python
# Schnittmenge: Mitarbeiter mit Fortbildung

df[df["MitarbeiterID"].isin(df2["MitarbeiterID"])][["Vorname","Nachname"]]


# Differenzmenge: Mitarbeiter ohne Fortbildung
df[~df["MitarbeiterID"].isin(df2["MitarbeiterID"])][["Vorname","Nachname"]]
```

```python
# inner join = Schnittmenge. Nur einträge die in beiden DFs vorkommen, wenn wir über 
# MitarbeiterID-Spalte mergen(verschmelzen)

pd.merge(df,df2, on="MitarbeiterID", how="inner")
```

```python
# left join = Verschmelzung aber alle zeilen aus linken DF bleiben erhalten.

pd.merge(df,df2, on="MitarbeiterID", how="left")
```

```python
# Outer Join 
# Es werden alle Mitarbeiter aus beiden DFs zusammengeführt -
# egal, ob sie nur in einer, in beiden oder in keiner Tabelle stehen.
# indicator = True erzeugt eine neue Spalte _merge, die folgendes zeigt:
# 1. "left_only": nur in df
# 2. "right_only": nur in df2
# 3. "both": in beiden
outerjoin = pd.merge(df,df2, on="MitarbeiterID", how="outer", indicator=True)
outerjoin
```

```python
# Übung
# df ist unser Personal-DataFrame
# Erstellen Sie ein sinnvolles zweites DataFrame 
# mit Mitarbeiterinnen, welche in Elternzeit sind
# Zusätzlich erstellen Sie eine Spalte in df mit restlichen
#  Urlaubstagen  
# Erhöhen Sie die Urlaubstage der IT-Abteilung um 2
# Erstellen Sie ein gemeinsames DataFrame, welches Informationen 
# liefert, ob die Mitarbeiter:innen in Elternzeit sind

df["Urlaubstage"] = df["MitarbeiterID"].apply(lambda x: random.randint(0,30))
print(df.loc[df["Abteilung"]=="IT","Urlaubstage"].head(1))
df.loc[df["Abteilung"]=="IT","Urlaubstage"]+=2 
print(df.loc[df["Abteilung"]=="IT","Urlaubstage"].head(1))
df_w = df[df["Geschlecht"]=="weiblich"]
elternzeit_ids = df_w["MitarbeiterID"].sample(1000, random_state = 1) 
# 1000 Mitarbeiterinnen in Elternzeit

elternzeit_list = []

for p in elternzeit_ids:
    eintrag = {
        "MitarbeiterID": p,
        "Beginn": fake.date_between(start_date="-1y", end_date="today"),
        "Ende": fake.date_between(start_date="today", end_date="+1y"),
        "Mutterschutzleistung": random.randint(1200,4000),
    }
    elternzeit_list.append(eintrag)

df3 = pd.DataFrame(elternzeit_list)
df3


outerjoin = pd.merge(df_w, df3, on="MitarbeiterID", how="outer", indicator=True)
outerjoin
```

# String-Methoden in pandas

```python
df["Beruf"].str.lower()
df["Beruf"].str.upper() 
df["Adresse"].str.title() # Anfangsbuchstabe jedes Wortes groß
df["Adresse"].str.capitalize() # Anfangsbuchstabe vom String
```

```python
df["Beruf"].str.replace("er","er:in") # alle Vorkommnisse von "er" durch "er:in" ersetzen
```

```python
df["Email"].str.len()
df["Email"].str.endswith(".de")
df["Abteilung"].str.find("I") # gibt die erste vorkommende Position sonst -1
df["Urlaubstage"]=df["Urlaubstage"].astype(str)
df["Urlaubstage"].str.zfill(2)

# Zählen Sie, wie viele Emailadressen mit example.com enden
# 1. Möglichkeit
df["Email"].str.endswith("example.com").sum()

# 2. Möglichkeit

len(df["Email"][df["Email"].str.endswith("example.com")])

# oder 
df["Email"][df["Email"].str.endswith("example.com")].count()
```

```python
# Erstellen Sie eine Spalte mit Initialien 
# z.b. Marc Schmidt -> MS
df["Initialien"] = df["Vorname"].str.get(0) + df["Nachname"].str.get(0)
df["Initialien"] = df["Vorname"].str[0] + df["Nachname"].str[0]
df["Initialien"]
```
