```python
from faker import Faker
import random 

fake = Faker("de_DE")

personen  = []

for _ in range(50000):
    person = {
        "name": fake.name(),
        "alter": random.randint(18,65),
        "adresse": {
            "stadt": fake.city(),
            "plz": fake.postcode()
        }
    }
    personen.append(person)

personen
```

# Übung
Schreiben Sie ein Programm, welches alle Personen unter 30 findet

```python
for person in personen:
    if person["alter"] < 30:
        print(person["name"], person["alter"])
```

Alle PLZ als Liste ohne Wiederholungen ausgeben

```python
plz_set = set([person["adresse"]["plz"] for person in personen])
plz_set
```
