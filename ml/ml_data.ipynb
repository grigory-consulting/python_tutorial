```python
import kagglehub
import pandas as pd
# Download latest version
path = kagglehub.dataset_download("pavansubhasht/ibm-hr-analytics-attrition-dataset")

print("Path to dataset files:", path)

import os

filepath = os.path.join(path, "WA_Fn-UseC_-HR-Employee-Attrition.csv")
df = pd.read_csv(filepath)
df.head()
```

```python
df["Attrition_Flag"] = (df["Attrition"] == "Yes").astype(int)
df["Attrition_Flag"] 
```

```python
features = ["Age", "MonthlyIncome", 'JobRole', 'DistanceFromHome', 'Gender', 'Department', "WorkLifeBalance", "RelationshipSatisfaction", "YearsAtCompany", "StandardHours"]

X = df[features].copy()  # Prädiktoren
y = df["Attrition_Flag"] # Regressor
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report


for col in ['JobRole', 'Gender', 'Department']:
    X[col] = LabelEncoder().fit_transform(X[col])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1)

clf = MLPClassifier()
clf.fit(X_train,y_train) # Modell-Training

y_pred = clf.predict(X_test) # Modell-Test
```

```python
accuracy_score(y_pred,y_test)
```

```python
print(classification_report(y_pred,y_test))
```
