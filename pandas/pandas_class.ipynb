```python
import pandas as pd
```

```python
class Patient:
    def __init__(self, first_name, last_name, birthday, height_cm, weight, bmi):
        self.first_name = first_name
        self.last_name = last_name
        self.birthday = birthday
        self.height_cm = height_cm
        self.weight = weight
        self.bmi = bmi 

    def __repr__(self):
        return f"Patient(vorname={self.first_name!r},nachname={self.last_name!r},))"
    
    def height_m(self):
        return self.height_cm / 100

    def is_tall(self):
        return self.height_cm >= 180
    
    def ist_normalgewicht(self):
        return (18<=self.bmi) and (self.bmi<=25)
```

```python
df = pd.read_pickle("patient.pk")
df["Patient_Obj"] = df.apply(
    lambda row:
    Patient(
    row["Vorname"], 
    row["Nachname"], 
    row["Geburtsdatum"], 
    row["Größe_cm"],
    row["Gewicht_kg"],
    row["BMI"],
    ),
    axis = 1 # zeilenweise entlang der Spalte
)

# Aufgabe: 
# 1. Erstellen Sie eine Spalte mit Größe in m
# 2. Listen Sie alle großen Patienten
# 3. Erstellen Sie eine Übersicht 
# für Durchschnittskörpergröße in m nach Geschlecht
# 4. Erweitern Sie die Klasse um Attribute Gewicht und BMI
# Erstellen Sie eine Klassenmethode ist_normalgewicht(18<=bmi<=25). 
# Wer sind die 12 jüngsten Patienten mit Normalgewicht.
```

```python
df["Größe_m"] = df["Patient_Obj"].apply(lambda p: p.height_m()) # 1.
df[df["Patient_Obj"].apply(lambda p: p.is_tall())] # Listen von großen Patienten
df.groupby("Geschlecht")[["Größe_m"]].mean()
```

```python
df[df["Patient_Obj"].apply(lambda p: p.ist_normalgewicht())].sort_values("Alter").head(12)[["Alter","Nachname"]]
```
